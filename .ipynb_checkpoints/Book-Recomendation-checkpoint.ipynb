# Collabritive Filtering Based Reccomender aproach !

```python
#Rember above we merge the book dataset and rating datase on basis of IBN(common) ! then this dataset created dont confuse 
ratings_with_name.head(5)
```

```python
#So we use dataset -> rating_with_name Which we created the above !

#1) Find user rate more than 200 books 
#2)-Find Books who have raitng more than 50 !
```

```python
ratings_with_name.groupby('User-ID')['Book-Rating'].count()
#so we see here the user -id and they telling how many book they rate 
#ex-> 17 mean 17 book 8user rate !

#So if you noteice we have 2Lakh+ user but only 90k User do rate the book ! (It happen ebc most of user not rate the book !)
```

```python
ratings_with_name.groupby('User-ID')['Book-Rating'].count() >200
#False mean not rate book more than 200 (mean they rate les book not rate 200 books )
```

```python
x = ratings_with_name.groupby('User-ID')['Book-Rating'].count()>200
p_user = x[x].index
#x[x] mean kept only true or positive value (and store there index of user)
```

```python
filtered_rating = ratings_with_name[ratings_with_name['User-ID'].isin(p_user)]
#simple we are saying (the user we get in p_user) is that presnet or same with User-Id if yes then kept only that user !
```

```python
#Now Filtering on the basis of the Books !
y = filtered_rating.groupby('Book-Title')['Book-Rating'].count() >=50
famous_books  = y[y].index
```

```python
famous_books
```

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
pt= final_ratings.pivot_table(index='Book-Title', columns='User-ID',values='Book-Rating')
pt.fillna(0,inplace=True)
pt
#Created pivot table Books more than 50 ratings and user rate more 200 ! books
#So we have 706 user(200 book rate !) and 810 books(books have 50! more rate )
```

# Finding the Realtion !

```python
from sklearn.metrics.pairwise import cosine_similarity
#So her elike each row will be the vector !
```

```python
#Now we sill find the cosine similarity for every row 
similarity_scores = cosine_similarity(pt)
similarity_scores
#so we are getting 2d graph here !
```

```python
#so this showing first row other all row relation(rmeber movie_recommend done !)
#if you notice at startng we have 1(mean same ,bec smae book same book ralation at starting !)
similarity_scores[0]
```

```python
#so we get by below using get the index of any book !
#np.where(pt.index=='Zoya')[0][0]
```

```python
#Now create the fucntion , we pass the book they reutr the 5 books to us !
def recommend(book_name):
    #fethc the index from the book name
    index = np.where(pt.index ==book_name)[0][0] # one i get index, now i go to the simialirty_scores(to find similarity with other)
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[1:6] 

    #one we get similar item now start the loop !
    for i in similar_items:
        print(pt.index[i[0]])
    
```

```python
recommend('1984') # checking by sending the book !
```
